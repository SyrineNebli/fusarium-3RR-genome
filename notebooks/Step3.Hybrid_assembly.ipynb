## Step 3: Hybrid Genome Assembly with SPAdes
**Input:** Merged Illumina reads (`02-primary/merged/`) 
and filtered Nanopore reads (`02-primary/trimmed/3RR-nanopore.fastq.gz`)  
**Output:** Primary hybrid assembly scaffolds in `04-assembly/spades/option-A/`  
**Tools:** SPAdes v4.2.0  
**Key parameters:** k-mer sizes: 25,33,37,47,59,71,91,111,127; 
`--isolate` mode for fungal genome optimization; 128 CPUs  
**Key finding:** 1,043 contigs; total length 42.86 Mb; N50 = 0.74 Mb; 
GC content 47.86%; average coverage 46×  
**Reference:** Materials & Methods Section 4.2 — Nebli et al. (2025)

# Configuration and Setup

```python
export SN=3RR
export NCPUS=128
```

#  Software Dependencies

```python
#Apptainer Aliases
alias spades="apptainer run docker://staphb/spades spades.py"
alias masurca_apptainer="apptainer run docker://staphb/masurca"
```

# Directory Structure Setup

```python
# Create main assembly directory
mkdir -p 04-assembly/{spades}

# Create subdirectories for different preprocessing options
mkdir -p 04-assembly/spades/{option-A}
```

# K-mer Size Selection

```python
# Define optimal k-mer sizes based on analysis (MAXIMUM=127)
export KMERS="25,33,37,47,59,71,91,111,127"
```

# SPAdes Hybrid Assembly

## Assembly Execution

```python
export opt=A

spades \
  -1 02-primary/merged/${SN}-${opt}-illumina_R1.fastq.gz \
  -2 02-primary/merged/${SN}-${opt}-illumina_R2.fastq.gz \
  -s 02-primary/merged/${SN}-${opt}-illumina.fastq.gz \
  --nanopore 02-primary/trimmed/${SN}-nanopore.fastq.gz \
  --isolate \
  -t $NCPUS \
  -k $KMERS \
  -o 04-assembly/spades/option-${opt}
```
